GridSearchCV optimization on top performing algorithms.

```python
import sys
sys.path.append('..')
from src.data_loader import load_or_generate_data
from src.preprocessing import clean_data, prepare_train_test_data, HRDataPreprocessor
from src.feature_engineering import add_engineered_features
from src.train import tune_best_model
from imblearn.over_sampling import SMOTE

df = add_engineered_features(clean_data(load_or_generate_data()))
X_tr, X_te, y_tr, y_te = prepare_train_test_data(df, target_col='Retention')
prep = HRDataPreprocessor()
X_tr_p = prep.fit_transform(X_tr)
y_tr_e = prep.fit_transform_target(y_tr)

smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(X_tr_p, y_tr_e)
best_tuned = tune_best_model(X_res, y_res, 'Random Forest')
print(best_tuned)
```
