# STEP 3: Exploratory Data Analysis (EDA)

Visualizing features against employee retention rates.

```python
import sys
sys.path.append('..')
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_loader import load_or_generate_data

df = load_or_generate_data()
plt.style.use('seaborn-v0_8-darkgrid')
```

```python
# Retention by Department
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='Department', hue='Retention', palette='viridis')
plt.title('Department-wise Retention Distribution')
plt.show()
```

```python
# Monthly Income Boxplot by Retention
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x='Retention', y='MonthlyIncome', palette='Set2')
plt.title('Monthly Income vs Retention')
plt.show()
```

```python
# OverTime vs Retention
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='OverTime', hue='Retention', palette='magma')
plt.title('OverTime Work vs Retention')
plt.show()
```
